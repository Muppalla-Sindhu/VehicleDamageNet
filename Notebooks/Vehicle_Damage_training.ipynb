Fetching models

```python
from ultralytics import YOLO

# Vehicle classifier (your trained model)
# Vehicle classifier
vehicle_classifier = YOLO(
    "/content/drive/MyDrive/Vehicle_Damage_Project/models/vehicle_classifier/best.pt"
)

# Damage models
car_damage_model = YOLO(
    "/content/drive/MyDrive/Vehicle_Damage_Project/models/car_damage_yolov8/weights/best.pt"
)

truck_damage_model = YOLO(
    "/content/drive/MyDrive/Vehicle_Damage_Project/models/truck_damage_yolov8/weights/best.pt"
)
bike_damage_model = YOLO(
    "/content/drive/MyDrive/Vehicle_Damage_Project/models/bike_damage_yolov8/weights/best.pt"
)

```

```python
def classify_vehicle(image):
    results = vehicle_classifier(image, verbose=False)
    probs = results[0].probs

    class_id = probs.top1
    class_name = results[0].names[class_id]

    return class_name.lower()

def detect_damage(image, vehicle_type):
    if vehicle_type == "car":
        model = car_damage_model
    elif vehicle_type == "truck":
        model = truck_damage_model
    elif vehicle_type == "bike":
        model = bike_damage_model
    else:
        return image, ["Vehicle type unknown – damage detection skipped"]

    results = model(image, conf=0.3, verbose=False)

    if results[0].boxes is None or len(results[0].boxes) == 0:
        return image, ["No visible damage detected"]

    annotated = results[0].plot()

    detected = set()
    for box in results[0].boxes:
        cls_id = int(box.cls[0])
        detected.add(model.names[cls_id])

    summary = [f"{dmg}" for dmg in detected]
    return annotated, summary
def vehicle_damage_pipeline(image):
    vehicle_type = classify_vehicle(image)

    annotated_img, damage_info = detect_damage(image, vehicle_type)

    final_text = f"Detected Vehicle: {vehicle_type.upper()}\n\n"
    final_text += "Damage Details:\n"
    final_text += "\n".join(damage_info)

    return annotated_img, final_text

```

```python
# =====================
# COST ESTIMATION RULES
# =====================
cost_table = {
    "CAR": {
        "Minor": (1000, 3000),
        "Moderate": (4000, 8000),
        "Severe": (15000, 40000)
    },
    "TRUCK": {
        "Minor": (2000, 5000),
        "Moderate": (7000, 15000),
        "Severe": (30000, 80000)
    },
    "BIKE": {
        "Minor": (500, 1500),
        "Moderate": (2000, 5000),
        "Severe": (8000, 20000)
    }
}
def estimate_cost(vehicle, damages):
    total_min = 0
    total_max = 0
    lines = []

    for d in damages:
        severity = severity_map.get(d, "Moderate")

        # fallback safety
        vehicle_key = vehicle if vehicle in cost_table else "CAR"

        min_c, max_c = cost_table[vehicle_key][severity]
        total_min += min_c
        total_max += max_c

        lines.append(f"- {d} → {severity} → ₹{min_c} – ₹{max_c}")

    return lines, total_min, total_max
if not damages:
    summary += "✅ No visible damage detected."
else:
    summary += "🛠 Damage Details:\n"
    for d in damages:
        severity = severity_map.get(d, "Unknown")
        summary += f"- {d} → {severity}\n"
```

```python
import gradio as gr
import cv2
import numpy as np
from ultralytics import YOLO

# =====================
# LOAD MODELS
# =====================
vehicle_model = YOLO(
    "/content/drive/MyDrive/Vehicle_Damage_Project/models/vehicle_classifier/best.pt"
)

damage_model = YOLO(
    "/content/drive/MyDrive/Vehicle_Damage_Project/models/car_damage_yolov8/weights/best.pt"
)

severity_map = {
    "Scratch": "Minor",
    "Dent": "Moderate",
    "Broken_Glass": "Moderate",
    "Wreck": "Severe"
}
# =====================
# COST ESTIMATION RULES
# =====================
cost_table = {
    "CAR": {
        "Minor": (1000, 3000),
        "Moderate": (4000, 8000),
        "Severe": (15000, 40000)
    },
    "TRUCK": {
        "Minor": (2000, 5000),
        "Moderate": (7000, 15000),
        "Severe": (30000, 80000)
    },
    "BIKE": {
        "Minor": (500, 1500),
        "Moderate": (2000, 5000),
        "Severe": (8000, 20000)
    }
}


# =====================
# VEHICLE CLASSIFIER
# =====================
def classify_vehicle(image):
    img = cv2.cvtColor(np.array(image), cv2.COLOR_RGB2BGR)
    results = vehicle_model.predict(img, verbose=False)

    if results and len(results[0].probs) > 0:
        class_id = int(results[0].probs.top1)
        return vehicle_model.names[class_id].upper()

    return None

# =====================
# DAMAGE DETECTOR
# =====================
def detect_damage(image):
    img = cv2.cvtColor(np.array(image), cv2.COLOR_RGB2BGR)
    results = damage_model.predict(img, conf=0.25, verbose=False)

    detected = set()
    annotated = img.copy()

    if results and results[0].boxes is not None:
        annotated = results[0].plot()
        for box in results[0].boxes:
            cls_id = int(box.cls[0])
            detected.add(damage_model.names[cls_id])

    annotated = cv2.cvtColor(annotated, cv2.COLOR_BGR2RGB)
    return annotated, detected
def estimate_cost(vehicle, damages):
    total_min = 0
    total_max = 0
    lines = []

    for d in damages:
        severity = severity_map.get(d, "Moderate")

        # fallback safety
        vehicle_key = vehicle if vehicle in cost_table else "CAR"

        min_c, max_c = cost_table[vehicle_key][severity]
        total_min += min_c
        total_max += max_c

        lines.append(f"- {d} → {severity} → ₹{min_c} – ₹{max_c}")

    return lines, total_min, total_max


# =====================
# MAIN PIPELINE
# =====================
def vehicle_damage_pipeline(image):
    vehicle = classify_vehicle(image)

    if vehicle is None:
        return image, "❌ No vehicle detected. Please upload a valid vehicle image."

    annotated, damages = detect_damage(image)

    summary = f"🚘 Detected Vehicle: **{vehicle}**\n\n"

    if not damages:
      summary += "✅ No visible damage detected.\n"
      summary += "💰 Estimated Repair Cost: ₹0"
    else:
      summary += "🛠 Damage Details:\n"
    for d in damages:
        severity = severity_map.get(d, "Unknown")
        summary += f"- {d} → {severity}\n"

    cost_lines, min_cost, max_cost = estimate_cost(vehicle, damages)

    summary += "\n💰 Repair Cost Estimate:\n"
    summary += "\n".join(cost_lines)
    summary += f"\n\n🔧 Estimated Total Cost: ₹{min_cost} – ₹{max_cost}"


    return annotated, summary

# =====================
# GRADIO UI
# =====================
with gr.Blocks(title="Multi-Vehicle Damage Detection") as demo:

    gr.Markdown(
        "## 🚗🏍️🚛 Multi-Vehicle Damage Detection System\n"
        "Upload an image to detect vehicle type and assess damage severity."
    )

    with gr.Row():
        with gr.Column():
            input_image = gr.Image(label="Upload Vehicle Image", type="pil")

        with gr.Column():
            output_image = gr.Image(label="Annotated Output")

    output_text = gr.Markdown(label="Detection Summary")

    with gr.Row():
        submit_btn = gr.Button("🔍 Analyze", variant="primary")
        clear_btn = gr.Button("🧹 Clear")

    submit_btn.click(
        fn=vehicle_damage_pipeline,
        inputs=input_image,
        outputs=[output_image, output_text]
    )

    clear_btn.click(
        fn=lambda: (None, ""),
        outputs=[input_image, output_text]
    )

demo.launch()
```
